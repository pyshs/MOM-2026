## Boucles

La liste `poids` contient le poids en grammes de morceaux de poteries

- Afficher chaque poids suivi de " g"
- Calculer le poids total du lot avec une boucle et une variable `total` qui commence à 0
- Afficher uniquement les poids supérieurs à 20g

```python
poids = [12, 45, 8, 30, 22, 5, 61, 17, 9, 28]
```

```python
for poids_poterie in poids:
    print(str(poids_poterie)  +          "g")
```

```python
[[poids_poterie, str(poids_poterie)+"g"] for poids_poterie in poids] #list comprehension
```

```python
liste_enrichie = [{"poids":poids_poterie, "affichage":str(poids_poterie)+"g"} for poids_poterie in poids] #list comprehension
liste_enrichie
```

```python
sum([element["poids"] for element in liste_enrichie])
```

```python
total = 0
for p in poids:
    total = total + p
total
```
