```python
sortie_fichier = True
nombre_min_lettres = 4

phrase = input("Écrivez une phrase : ")
liste_mots = phrase.split()

if len(liste_mots) == 0:
    print("Vous n'avez rien écrit")
else:
    nombre_mots = len(liste_mots)
    compteur = 0
    for mot in liste_mots:
        mot_nettoye = mot.strip(".,;:!?«»\"'()")
        if len(mot_nettoye) >= nombre_min_lettres:
            compteur += 1

    proportion = round(100 * compteur / nombre_mots, 2)
    informations = {
        "Total": compteur,
        "Proportion": proportion,
        "Phrase": phrase,
        "Seuil": nombre_min_lettres,
    }

    print(f"Proportion de mots avec {nombre_min_lettres} lettres ou plus : {proportion} %")

    if sortie_fichier:
        with open("resultat.txt", "w", encoding="utf-8") as f:
            for cle, valeur in informations.items():
                f.write(f"{cle} : {valeur}\n")
```

Je veux isoler le bloc qui prend un texte et calcul le nombre de mots de plus d'un certain seuil de caractè!res. Je vais l'appeler `calcul_mots_long`

- en entrée prend un texte et un seuil
- fait des opérations
- retourne le nombre de mots de plus d'un certain seuil

```python
def calcul_mots_long(phrase, nombre_min_lettres = 4):
    """
    Prend en entrée un texte phrase et renvoie le nombre de mots longs
    """
    liste_mots = phrase.split()
    
    if len(liste_mots) == 0:
        print("Vous n'avez rien écrit")
        return None
    else:
        nombre_mots = len(liste_mots)
        compteur = 0
        for mot in liste_mots:
            mot_nettoye = mot.strip(".,;:!?«»\"'()")
            if len(mot_nettoye) >= nombre_min_lettres:
                compteur += 1
    
        proportion = round(100 * compteur / nombre_mots, 2)
        informations = {
            "Total": compteur,
            "Proportion": proportion,
            "Phrase": phrase,
            "Seuil": nombre_min_lettres,
        }
        return informations
```
