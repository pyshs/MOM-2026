# Mon premier notebook

Exécuter notre script Python

````python
sortie_fichier = False # Déclaration de variable
nombre_min_lettres = 5 # Déclaration de variable
phrase = input("Écrivez une phrase : ") # On demande une entrée clavier à l'utilisateur
liste_mots = phrase.split() # On coupe la chaine de caractères aux espaces
if len(liste_mots) == 0: # On test si le nombre d'élément dans la liste de mots est égal à 0
    print("Vous n'avez rien écrit") # On écrit s'il n'y a pas de mots
else: # Sinon on continue avec le traitmeent 
    nombre_mots = len(liste_mots) # on créer une nouvelle variable du nombre de mots
    compteur = 0 # on définit une variable à 0
    for mot in liste_mots: # boucle sur la liste des mots
        mot_nettoye = mot.strip(".,;:!?«»\"'()") # nettoyer pour chaque mot les ponctuations
        if len(mot_nettoye) >= nombre_min_lettres: # on vérifie la condition si la longueur du mot est plus grande que le seuil
            compteur += 1 # si la condition est vérifiée, on incrémente le compteur compteur = compteur + 1
    proportion = round(100 * compteur / nombre_mots, 2) # Calculer la proportion
    informations = {
        "Total": compteur,
        "Proportion": proportion,
        "Phrase": phrase,
        "Seuil": nombre_min_lettres,
    }
    print(f"Proportion de mots avec {nombre_min_lettres} lettres ou plus : {proportion} %")
    if sortie_fichier: # Condition d'écriture dans un fichier
        with open("resultat.txt", "w", encoding="utf-8") as f:
            for cle, valeur in informations.items():
                f.write(f"{cle} : {valeur}\n")
````
