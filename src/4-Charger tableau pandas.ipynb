On veut catégoriser les dates entre
- avant les années 2000
- après 2000

Comment faire ?

- On va faire une fonction qui catégorise une date
- Et on va la passer à tout le tableau

```python
def recoder_date(date):
    if pd.isna(date):
        return None
    if int(date) < 2000:
        return "avant 2000"
    else:
        return "après 2000"

recoder_date(2005)
```

```python
df["Année"].apply(recoder_date)
```

```python
df["Année_recodée"] = df["Année"].apply(recoder_date)
```
